```python
import cv2
from ultralytics import YOLO
import numpy as np
from pathlib import Path

# --- Configuration ---
# make sure this path matches where your ONNX model was saved
onnx_model_path = Path("./artifacts/models_saved/agrivision_maladies_s_v1/agrivision_maladies_s.onnx")

# path to a sample image for inference
# you'll need to replace this with an actual image path you want to test
sample_image_path = "path/to/your/sample_image.jpg" 

# --- Load the ONNX model ---
if not onnx_model_path.exists():
    print(f"🚨 error: onnx model not found at {onnx_model_path}")
    print("please ensure you've run collect_artifacts.py and the path is correct.")
else:
    print(f"✅ loading onnx model from: {onnx_model_path}")
    model = YOLO(str(onnx_model_path)) # load the exported onnx model

    # --- Perform Inference ---
    if not Path(sample_image_path).exists():
        print(f"🚨 error: sample image not found at {sample_image_path}")
        print("please update 'sample_image_path' to a valid image file.")
    else:
        print(f"✨ performing inference on: {sample_image_path}")
        results = model(sample_image_path) # predict on the image

        # --- Process and Display Results ---
        for r in results:
            im_array = r.plot()  # plot results, bounding boxes, etc.
            im_rgb = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB) # convert to rgb for matplotlib

            # display image with matplotlib if in a notebook
            try:
                import matplotlib.pyplot as plt
                plt.figure(figsize=(10, 8))
                plt.imshow(im_rgb)
                plt.axis('off')
                plt.title(f"predictions for {Path(sample_image_path).name}")
                plt.show()
            except ImportError:
                print("matplotlib not found, saving image with detections instead.")
                # fall back to saving if matplotlib isn't available
                output_image_name = f"predictions_{Path(sample_image_path).name}"
                cv2.imwrite(output_image_name, im_array)
                print(f"detections saved to {output_image_name}")

            print("\n--- detection details ---")
            print(f"boxes: {r.boxes}") # bounding box details
            print(f"masks: {r.masks}") # segmentation masks (if any)
            print(f"keypoints: {r.keypoints}") # keypoints (if any)
            print(f"probs: {r.probs}") # class probabilities (for classification models)
```
